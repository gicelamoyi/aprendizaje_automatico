##1. importo libreria

```python

import pandas as pd
import numpy as np
from sklearn import linear_model
from sklearn import model_selection
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

##2.Cargar dataset

```python
df = pd.read_csv('usuarios_win_mac_lin (1).csv')

print("--- VISTA PREVIA DEL DATASET ---")
print(df.head())
```

##3.Separar variables y dividir en entrenamiento/prueba

```python
X = df.iloc[:, :4].values
y = df.iloc[:, 4].values

X_train, X_test, y_train, y_test = model_selection.train_test_split(
    X, y, test_size=0.20, random_state=42
)
```

##4.Crear y entrenar el modelo

```python
clasificador = linear_model.LogisticRegression(solver='lbfgs', max_iter=1000)
clasificador.fit(X_train, y_train)
```

##5.Predecir sobre conjuto de prueba

```python
y_pred = clasificador.predict(X_test)
```

##6.evaluar modelo

```python
print("\n--- RESULTADOS DE EVALUACIÓN ---")
print("Exactitud (Accuracy):", round(accuracy_score(y_test, y_pred), 4))

print("\n--- MATRIZ DE CONFUSIÓN ---")
print(confusion_matrix(y_test, y_pred))

print("\n--- REPORTE DE CLASIFICACIÓN ---")
print(classification_report(y_test, y_pred, target_names=['Windows', 'Macintosh', 'Linux']))
```

##7. predecir un nuevo usuario

```python
nuevo_usuario = np.array([[120, 5, 18, 35]])
prediccion = clasificador.predict(nuevo_usuario)

sistemas = {0: 'Windows', 1: 'Macintosh', 2: 'Linux'}
print(f"\nEl sistema operativo predicho es: {sistemas[prediccion[0]]}")
```
