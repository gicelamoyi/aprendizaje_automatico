# Ejercicio 2 - Visualización de datos con Matplotlib

**Objetivo:** inspeccionar visualmente el dataset de automóviles (`Automobile.csv`) para identificar
patrones, distribuciones, relaciones y posibles outliers.

> **Nota sobre el dataset:** el archivo `Automobile.csv` usado acá es un dataset **generado sintéticamente**
> con la misma estructura de columnas que el dataset original de Kaggle
> (mpg, cylinders, displacement, horsepower, weight, acceleration, model_year, origin).
> Si tu profesor pide el dataset real, descargalo de Kaggle y reemplazá el archivo en `../datos/Automobile.csv`
> manteniendo el mismo nombre y columnas.


```python
import pandas as pd
import matplotlib.pyplot as plt

ruta_datos = "../datos/Automobile.csv"
```

## 1. Cargar el conjunto de datos

```python
df = pd.read_csv(ruta_datos)
print("Filas y columnas:", df.shape)
df.head()
```

## 2. Gráfico de dispersión: mpg vs weight

```python
plt.figure(figsize=(8,5))
plt.scatter(df["weight"], df["mpg"], alpha=0.7, color="steelblue")
plt.title("Relación entre peso y consumo de combustible")
plt.xlabel("Peso (weight)")
plt.ylabel("Millas por galón (mpg)")
plt.grid(True, alpha=0.3)
plt.show()
```

## 3. Histograma de mpg

```python
plt.figure(figsize=(8,5))
plt.hist(df["mpg"], bins=15, color="darkorange", edgecolor="black")
plt.title("Distribución del consumo de combustible (mpg)")
plt.xlabel("mpg")
plt.ylabel("Frecuencia")
plt.show()
```

## 4. Boxplot de cylinders y origin

```python
fig, axes = plt.subplots(1, 2, figsize=(10,5))
axes[0].boxplot(df["cylinders"])
axes[0].set_title("Boxplot: cylinders")
axes[1].boxplot(df["origin"])
axes[1].set_title("Boxplot: origin")
plt.tight_layout()
plt.show()
```

## 5. Gráfico de barras: cantidad de autos por model_year y origin

```python
tabla = df.groupby(["model_year", "origin"]).size().unstack(fill_value=0)
tabla.plot(kind="bar", figsize=(10,5))
plt.title("Cantidad de automóviles por año de modelo y origen")
plt.xlabel("Año de modelo")
plt.ylabel("Cantidad de autos")
plt.legend(title="Origen (1=EEUU, 2=Europa, 3=Asia)")
plt.tight_layout()
plt.show()
```

## Conclusión

Con estos gráficos exploramos relaciones entre variables (mpg vs weight), la distribución del consumo,
posibles valores atípicos en cilindros y origen, y la composición del dataset por año y origen.
