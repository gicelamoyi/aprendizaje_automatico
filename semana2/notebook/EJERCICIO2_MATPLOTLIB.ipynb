```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Automobile.csv')
df.head()
```

```python
plt.scatter(df['weight'], df['mpg'])
plt.xlabel('Peso (weight)')
plt.ylabel('Millas por galón (mpg)')
plt.title('Relación entre peso y consumo')
plt.show()
```

```python
plt.hist(df['mpg'], bins=15)
plt.xlabel('mpg')
plt.ylabel('Frecuencia')
plt.title('Distribución del consumo de combustible')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10,5))
axes[0].boxplot(df['cylinders'])
axes[0].set_title('Cylinders')
axes[1].boxplot(df['origin'])
axes[1].set_title('Origin')
plt.show()
```

```python
tabla = df.groupby(['model_year', 'origin']).size().unstack()
tabla.plot(kind='bar', figsize=(10,5))
plt.xlabel('Año de modelo')
plt.ylabel('Cantidad de autos')
plt.title('Autos por año y origen')
plt.show()
```
