# Ejercicio 1 - Importación y manipulación de datos con Pandas

**Objetivo:** cargar datos de ventas (CSV), clientes (JSON) e inventario (Excel), y realizar operaciones básicas de análisis.

Archivos usados (carpeta `../datos/`):
- `ventas.csv`
- `clientes.json`
- `inventario.xlsx`


```python
import pandas as pd

# Rutas relativas a la carpeta datos/
ruta_ventas = "../datos/ventas.csv"
ruta_clientes = "../datos/clientes.json"
ruta_inventario = "../datos/inventario.xlsx"
```

## 1. Cargar datos de ventas (CSV) y ver una vista previa

```python
df_ventas = pd.read_csv(ruta_ventas)
print("Filas y columnas:", df_ventas.shape)
df_ventas.head()
```

## 2. Cargar datos de clientes (JSON) y contar cuántos hay

```python
df_clientes = pd.read_json(ruta_clientes)
cantidad_clientes = df_clientes.shape[0]
print(f"Cantidad total de clientes: {cantidad_clientes}")
df_clientes.head()
```

## 3. Cargar datos de inventario (Excel) y calcular el stock promedio

```python
df_inventario = pd.read_excel(ruta_inventario)
promedio_stock = df_inventario["stock_disponible"].mean()
print(f"Promedio de stock disponible: {promedio_stock:.2f} unidades")
df_inventario.head()
```

## Conclusión

En este notebook importamos datos de tres fuentes distintas (CSV, JSON, Excel) usando Pandas,
y calculamos indicadores básicos: vista previa de ventas, cantidad de clientes y stock promedio.
